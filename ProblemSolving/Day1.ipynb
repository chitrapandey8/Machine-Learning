```python
import math

def Prime(n):
    cnt = 0
    for i in range(1,n+1):
        if(n%i == 0):
            cnt = cnt+1


    return cnt        


n = int(input("Enter:"))
if Prime(n) == 2:
    print("Prime")
else:
    print("Not Prime")    

```

```python
def prime(n):
    cnt = 0
    i = 1
    while i*i <= n:
        if n%i == 0:
            cnt += 2
        i = i+1

    return cnt == 2



n = int(input("E"))
if prime(n):
    print("prime")
else:
    print("not prime")            
```

```python
n = int(input())
def squrtt(n):
    i = 1
    while i*i <= n:
        i = i+1
    return i-1
    

ans = squrtt(n)
print(ans)   ##square root of n iterations    
```

```python
n = int(input(10))
def sqrt(n):
    low = 1
    end = n
    ans = 0
    while low <= end:
        mid = low + (end - low)//2

        if mid*mid == n:
            return mid

        if mid*mid > n:
            end = mid-1
        else:
            ans = mid
            low = mid+1   
        
    return ans

ans = sqrt(n)
print(ans)
```
