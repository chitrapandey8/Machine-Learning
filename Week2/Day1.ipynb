### Function Kwargs

```python
#kwars if for keyword argument:
def Person(name, age, professsion, **kwargs):
    print(name, age, professsion)
    print(kwargs)


Person("nmeh",44, "e3e3e3", city="uhdfd", h=2)

 ##kwargs is a dictionary, kwargs must be at end, kwargs keyword argumwnts ke liye hota hai

```
