##What is a function?

##A function is a block of code which only runs when it is called.
##In Python a function is defined with the def keyword.
##Functions can accept any number of arguments of any type.
##Functions can optionally return data as well. This data may also ##be of any type. You can return more than one value as well.
##Arguments can have default values as well.
##While calling functions, the user may define arguments as keyword ##pairs.

```python
def hello():
    print("hello")


def sum(a,b):
    return a+b
   
```

### MAP FUNTIONS

```python
##Generate a list that contains the squares of the numbers in 'a'

a = [1,2,3,4,5]
b = []

for i in a:
    b.append(i ** 2)
print(b)
```

```python
c = list(map(Square,a)) ## map fucntion is used for iteration over a list, it takes two argumentrs a name of the function, and a iteratable
print(c) ##map return map object then we convert it into list
```
