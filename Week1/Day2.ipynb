# Breaks and Continue

```python
for i in range(1,5):
    if i == 3:
        break
    print(i)
        
```

```python
for i in range(1,5):
    if(i == 3):
        continue
    print(i)
```

### Random Game

```python
import random

while True:
    r = random.randint(1,10)

    if r == 10:
        break

    print(r, end=" ")
```

Python Lists

Lists are used to store multiple items in a single variable.
Lists are created using square brackets.
List items are ordered, changeable, and allow duplicate values.
List items are indexed, the first item has index [0], the second item has index [1], etc.
The list is changeable, meaning we can change, add, and remove items in a list after it has been created.
You can also use the list constructor list() to create lists.

```python
##List in python is Heterogenaous
a = [2,4,56,66]
print(a)

aa = ["i", "am", "chitarsnhi"]
print(aa)

hello = [[3],"hurhf",True,33.33]
print(hello)
```

```python
##Length of String
a = [3,4,6,8,6,6,7]
print(len(a))

#iteration in List
for i in a:
    print(i,end=" ")

print()

#accessing by index
print(a[0])

##Swapping in List
a[0],a[1]  = a[1],a[0]
print(a)

a.insert(3,"hello")
print(a)

##pop will also show waht popped off
x = a.pop()
print(a)
print(x)

a.append(7)
print(a)


##Creating List
a = []
a = list()

list(range(2,10))
```

#### Linear Search

```python
a = [24,57,45,6,45,6,3]

#One Way
if 6 in a:
    print("yes")

##Another Way
b = 6
for i in range(0,len(a)):
    if(a[i] == b):
        print(i)
        break
```
