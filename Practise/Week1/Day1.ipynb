```python
##Rotate List by K:
def rotate(k, a =[]):
    ans = []
    aa = []
    l1 = a[k-1::-1]
    l2 = a[:k-1:-1]
    for i in l1:
        ans.append(i);
    for i in l2:
        ans.append(i)

    aa =  ans[len(a)::-1]
    return aa
    

ans = []
ans = rotate(3,[1,2,3,4,5,6])
print(ans)


```
