```python
##list intersection without sets
list1 = [1,2,3,4,5]
list2 = [3,4,5,6,7]


ans = []
for i in range(len(list1)):
    if list1[i] in list2:
        ans.append(list1[i])

print(ans)        
    
```

```python
#Running sum:
l  = [1,2,3,4]
presum = []

presum.append(l[0])
for i in range(1,len(l)):
    presum.append(presum[i-1]+l[i])


print(presum)    


```

```python
#Move zeroes to end
l = [0,1,0,3,12]
ans = []
for a in l:
    if a != 0:
        ans.append(a)


for a in l:
    if a == 0:
        ans.append(a)

print(ans)       
```

```python
def swapfirstandlast(t = ()):
    l = list(t)
    first = l[0]
    second = l[len(l)-1]
    l[0] =  second
    l[len(l)-1] = first
    return tuple(l)
    
    

tt = (1,2,3,4)
ans = swapfirstandlast(tt)
print(ans)
```
