```python
##Remove Duplicates Preserving Order:
l = [4,2,3,2,1,4,5]
s = set()

ans = []

for x in l:
    if x not in s:
        ans.append(x)
        s.add(x)
    else:
        continue    

        
print(ans)        
```

```python
#Set of Primea up to N:
n = 50

ans = []
for i in range(1,n+1):
    prime = True
    for x in range(2,i):
        if i % x == 0:
            prime = False
    
    if prime:
        ans.append(i)
            

        


print(ans)
```

```python
##Word Frequency:
s = "hello world hello python world hello"
l = s.split()

d = dict()

for x in l:
    if x in d:
        d[x] += 1
    else:
        d[x] = 1


print(d)        
```

```python
d = {"a":1, "b":2,"c":1}
ans = dict()
for x in d.items():
    if x[1] not in ans:
        l = list()
        l.append(x[0])
        ans[x[1]] = l
    else:
        ans[x[1]].append(x[0])


print(ans)    
```

```python
##Merge Two Dictonaries:
d1  = {"a":1, "b":2}
d2  = {"b":3, "c":4}
ans = dict()


for i in d1.items(): ## ye tuple mai rakhta hai
    if i[0] not in ans.keys():
        ans[i[0]] = i[1]
    else:
        continue


for i in d2.items(): ## ye tuple mai rakhta hai
    if i[0] not in ans.keys():
        ans[i[0]] = i[1]
    else:
        no = ans[i[0]]
        no += i[1]
        ans[i[0]] = no         



print(ans)
```

```python
##Group by first charcter

l = ["apple","apricot", "banana", "blueberry", "cherry"]
d = dict()
for i in l:
    if i[0] not in d.keys():
        l = list()
        l.append(i)
        d[i[0]] = l
    else:
        d[i[0]].append(i)    



print(d)
```
