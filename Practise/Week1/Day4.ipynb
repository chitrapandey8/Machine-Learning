```python
#Max Value Key::
d = {"math":85, "science":92, "english":78, "history":92}
ans = []
maxvalue = max(d.values())
for i,j in d.items():
    if j == maxvalue:
        ans.append(i)



print(ans)

```

```python
##Nested Dictionary Traversal::
students = {
"Alice": {"scores": [85,90,78], "grade": "B"},
"Bob": {"scores": [92,88,94], "grade": "A"}
}

ans = dict()

for i,j in students.items():
    for k,l in j.items():
        if k == "scores":
            summ = sum(l)
            ans[i] = summ
            
print(ans)            
```

```python
##Dictionaries to list of tuples:
d =  {"banana":3, "apple":5, "cherry":2}
ans = list()


for x in d.items():
    ans.append(x)

print(ans)    
```

```python
##simple calculator:
num1 = int(input("enter num1"))
num2 = int(input("Enter num2"))
op = input("Enter operator")
ans = 0

if op == "+":
    ans =  num1+num2
    print(ans)
elif op == "-":
    ans = num1-num2
    print(ans)
elif op == "*":
    ans = num1*num2
    print(ans)
elif op == "/":
    if num2 == 0:
        print("cannot be divided")
    else:
        ans = num1/num2
        print(ans)
elif op == "%":
    ans = num1%num2    
    print(ans)
else:
    print("Invalid op")        


        
            
```
