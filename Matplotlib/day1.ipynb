```python
##creating multiple plot on y axis
x_val = np.arange(-3,3.1,0.1)
y_val = x_val **2
y_val2 = x_val ** 3
 
plt.plot(x_val, y_val)
plt.plot(x_val, y_val2, "y*")

plt.show()
```

```python
##creating multiple plot on y axis
x_val = np.arange(-3,3.1,0.1)
y_val = x_val **2
y_val2 = x_val ** 3
 
plt.plot(x_val, y_val, label = "Square")
plt.plot(x_val, y_val2, label = "Cube")
plt.legend()

plt.show()
```

```python
##creating multiple plot on y axis
x_val = np.arange(-3,3.1,0.1)
y_val = x_val **2
y_val2 = x_val ** 3
 
plt.plot(x_val, y_val)
plt.plot(x_val, y_val2)
plt.legend(["Square", "cube"], loc = "lower right", frameon = False) ##loc for location 

plt.show()
```

```python
##creating multiple plot on y axis
x_val = np.arange(-3,3.1,0.1)
y_val = x_val **2
y_val2 = x_val ** 3
 
plt.plot(x_val, y_val)
plt.plot(x_val, y_val2)
plt.legend(["Square", "cube"], loc = "lower right",ncol = 2 ,frameon = False) ##loc for location 

plt.show()
```

```python
##creating multiple plot on y axis
x_val = np.arange(-3,3.1,0.1)
y_val = x_val **2
y_val2 = x_val ** 3
 
plt.plot(x_val, y_val)
plt.plot(x_val, y_val2)

plt.text(-1,0+1,"(0,0)")
plt.text(1,1+2,"(1,1)")
plt.grid()
plt.legend(["Square", "cube"], loc = "lower right",ncol = 2 ,frameon = False) ##loc for location 

plt.show()
```
