```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(20)

y1 = np.linspace(1, 20, 20) + np.random.rand(20) * 5
y2 = np.linspace(1, 20, 20) + np.random.rand(20) * 10

plt.scatter(x, y1)
plt.scatter(x, y2)

plt.legend(["Group-1", "Group-2"])

plt.show()
```

```python


x = np.arange(20)

y1 = np.linspace(1, 20, 20) + np.random.rand(20) * 5
y2 = np.linspace(1, 20, 20) + np.random.rand(20) * 10

plt.scatter(x, y1, s=y1+5)
plt.scatter(x, y2, s=y2+5)

plt.legend(["Shop-1", "Shop-2"])

plt.show()
```

```python

##Sigmoid fucntion//Logistic function
import math
Sigmoid = lambda x:1/(1+math.exp(-x))
x = np.arange(-100, 100)
y = [Sigmoid(i) for i in x]
plt.plot(x,y)
plt.show(-10, 10)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 10, 0.1)

y1 = np.sin(x)
y2 = np.tan(x)
y3 = np.log(x)

plt.figure()

plt.subplot(3, 1, 1)
plt.plot(x, y1)

plt.subplot(3, 1, 2)
plt.plot(x, y2)

plt.subplot(3, 1, 3)
plt.plot(x, y3)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 10, 0.1)

y1 = np.sin(x)
y2 = np.tan(x)
y3 = np.log(x)

fig = plt.figure()

ax231 = fig.add_subplot(231)
ax231.plot(x, y1)

ax233 = fig.add_subplot(233)
ax233.plot(x, y2)

ax235 = fig.add_subplot(235)
ax235.plot(x, y3)

plt.show()
```
