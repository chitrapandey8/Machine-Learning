```python
arr = [2,7,11,17]
target = 9
d = dict()
ans = {}
for i in range(len(arr)):
    diff = target - arr[i]
    if diff in d:
        ans[d[diff]] = i
    if arr[i] not in d:
        d[arr[i]] = i   

    


print(ans)
```

```python
##MAximum Absolute difference:
A = [1,3,-1]
ans = 0
for i in range(len(A)):
    for j in range(len(A)):
        x = abs(A[i]-A[j]) + abs((i+1)-(j+1))
        ans = max(ans,x)
print(ans)
```

```python
##
A = [1,3,-1]  ## find max of this function |A[i]-A[j]| + |i-j|

inf = 10 ** 9
a = -inf
b = inf
c = -inf
d = inf

for i in range(len(A)):
    x = A[i]

    a = max(a,x+i)
    b = min(b,x+i)
    c = max(c,x-i)
    d = min(d,x-i)

ans = max(a-b, c-d)


print(ans)
```

```python
##first missing integer
arr = [3,4,-1,1]

def firstmissingpos(arr):
    n = len(arr)
    for i in range(len(arr)):
        while arr[i] > 0 and arr[i] <= n and arr[arr[i]-1] != arr[i]:
            index_to_swap = arr[i]-1
            arr[i], arr[index_to_swap] = arr[index_to_swap], arr[i]

    ans = 0
    for i in range(len(arr)):
         if arr[i] != i+1:
             ans = i+1
    return ans         
             
             


ans = firstmissingpos(arr)
print(ans)

    
```
