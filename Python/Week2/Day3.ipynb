```python
def Decimaltobianry(n):
    l = []
    while n > 0:
        remainder = n%2
        l.append(str(remainder))
        n = n//2
    l.reverse()
    s = "".join(l)    #//join array of string pe lgta hia
    return int(s)

Decimaltobianry(10)    
 
```

```python

def deci(n):
    pow = 1
    ans = 0
    while n!=0:
        rem = n%2
        ans = ans+rem*pow
        pow = pow*10
        n = n//2
    return ans    

ans = deci(10)
print(ans)
```

```python
def BtoD(n):
    pow = 1
    ans = 0
    while n!=0:
        rem = n%10
        ans = ans + pow*rem
        pow = pow*2
        n = n//10
    return ans   

    
ans = BtoD(1010)
print(ans)
```

```python
##Trailing Zeros:
def TrailingZeros(n):
    i = 5
    ans = 0
    while i<=n:
        ans += n//i
        i = i*5   ##1,2,3,4,5,6,7,10,15,20,25, diving by 5 will see the contribution till 20 but 25 contribute 2 fives
    return ans;


ansc = TrailingZeros(30)
print(ansc)

```
