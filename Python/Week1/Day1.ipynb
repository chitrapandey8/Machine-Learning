# LOOPs

```python
for i in range(10,0,-1):
    if i%2 != 0:
        print(i, end=" ")

print(end=" ")
for i in range(9, 0, -2):
    print(i, end=" ")        
```

```python
i = 234
sum = 0
while(i):
    last = i%10
    sum += last
    i = i//10

print(sum)    
```

```python
for i in range(0,5):
    for j in range(1,i+1):
        print(j, end=" ")
    
    print()    
```

```python
n = 4
for i in range(1,5):
        print("0"*(n-i), end="")

        for j in range(i, 2*i):
         print(j, end="")
        

        for j in range(2*i-2,i-1, -1):
          print(j, end="") 

        print("0"*(n-i), end="")

        print()
        
```
