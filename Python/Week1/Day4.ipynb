Python Lists
Lists are used to store multiple items in a single variable.
Lists are created using square brackets [].
List items are ordered, changeable, and allow duplicate values.
List items are indexed:
The first item has index 0
The second item has index 1
and so on.
Lists are changeable, meaning we can change, add, and remove items after the list has been created.
You can also use the list constructor list() to create lists.

```python
l = [1,2,3,4,5,6,7]
```

Tuples
Tuples are used to store multiple items in a single variable.
A tuple is a collection which is ordered and immutable (unchangeable).
Tuples are written with round brackets ().
Tuple items are indexed:
The first item has index [0]
The second item has index [1]
and so on.
Tuples can be created with the tuple constructor tuple().
Tuples can also be created by simply writing multiple values separated by commas.
Tuples are processed faster than lists in Python.

```python
##Tuples --- tuples are readonly list
v = (1,2,3,4,54)
print(v)
```

```python
a = tuple() ## these are constructors
a = tuple([2,4,5,3])
print(a)
```

```python
def product(*args):  ##args is a tuple of argument, we can also iterate over it
    p = 1
    for n in args:
        p = p*n
    return p  

product(1,2,3)
```

### SETS

Sets
A set is an unordered collection of items, where every element is unique and immutable. However, the set itself is mutable.
There are two ways of creating a set:
Putting all the elements inside curly braces {}.
Using the set() method.
Adding elements to a set: add(), update().
Removing elements from a set: remove(), discard().
Python sets allow fast insertion, deletion, and searching.

```python
a ={1,2,3,4,5,5,6,-1,1,-3,"hello"}
print(a)
```

```python
s = "abaa"
mp = {}
for char in s:
    if char in mp:
        mp[char] += 1
    else:
        mp[char] = 1


odd_freq = 0
for freq in mp.values():
    if freq%2 == 1:
        odd_freq += 1


print(mp)

if odd_freq <= 1: ##agr odd kese bhi ek se jayada baar aata hi toh not possible
    print("yas")
else:
    print("no")            

    
```
