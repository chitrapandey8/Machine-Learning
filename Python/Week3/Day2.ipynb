```python
def mul(x):
    return x*2

arr = [1,2,3]
list2 = list(map(mul,arr))
print(list2)
```

```python
##filter
list2 = [1,2,4,5]
list3 = list(filter(lambda x:x%2 == 0, list2))
print(list3)
```

```python
##reduce
import functools
list2 = [1,2,3,4,5]
num = functools.reduce(lambda a,b: a+b, list2)
print(num)
```

```python
arr = [1,2,3,4.5,-1,0]

list1 = list(filter(lambda x: x > 0, arr))
print(list1)
list2 = list(map(lambda x: x*2.54, list1))
print(list2)
result = functools.reduce(lambda x,y:x+y, list2)
print(result)
```
