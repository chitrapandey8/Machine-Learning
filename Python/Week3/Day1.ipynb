```python
class Human():  ###Encapsulation means putting properties and methods togethher in a single entity 
    def __init__(self, name, age, Gender):
        self.name = name
        self.age = age
        self.Gender = Gender

    def Myself(self):
        print(f"Hello Im {self.name},my age is {self.age}, my gender is {self.Gender}")    
        

        
```

```python
class Animal:
    def __init__(self, name):
        self.name = name
        

    def Property(self):
        print("I BArk")

        

class Dog(Animal):
    def __init__(self, name,age):
        super().__init__(name)
        self.age = age


    def Describe(self):
        print(f"I am {self.name} age is {self.age}")

        
D = Dog("DOG", 2)
D.Describe()
D.Property()

```

```python
##Abstraction --user function call kar ra or uske piche kya functionality hai voh nhi bataungi

from abc import ABC,abstractmethod

class car(ABC):
    @abstractmethod
    def start(self):
        pass


class Car(car):
    def start(self):
        print("IM A CAR")
        


c = Car()
c.start()

```
