```python
class acc:
    next_id = 0
    def __init__(self, balance):
        acc.next_id += 1
        self.id = acc.next_id
        self.balance = balance

    def des(self):
        print(f"acc: {self.id}, bal: {self.balance}" )    
    
    def desposit(self,amount):
        self.balance += amount

    def Withdraw(self, amount):
        self.balance -= amount 

    def __regr__(self):
        return {self.id}    
        
```

```python
acc1 = acc(1000)
acc1.des()
acc1.desposit(100)
acc1.des()
acc1.Withdraw(200)
acc1.des()
acc1.__regr__()
```

## Inheritance

```python
class acc:
    next_id = 0
    def __init__(self, openbalance = 0):
        acc.next_id += 1
        self.id = acc.next_id
        self.balance = openbalance
        self.num_transactions = 0
        self.max_transactions = 5

    def des(self):
        print(f"acc: {self.id}, bal: {self.balance}" )    
    
    def desposit(self,amount):
        if amount>=0 and self.num_transactions < self.max_transactions:
            self.balance += amount
            self.num_transactions += 1

    def Withdraw(self, amount):
        if amount >= 0 and self.balance >= amount and self.num_transactions < self.max_transactions:
            self.balance -= amount 
            self.num_transactions += 1

    
    def __str__(self):
        return f"{self.id} {self.balance}"

    def __regr__(self):
        return {self.id}   
```

```python
class SavingsAcc(acc):
    pass

```

```python
class CurrentAcc(acc):
    def __init__(self, openbalance=0): ##it call init of parent class then go further
        super().__init__(openbalance)  
        self.max_transactions = 10 ##Parent ki max_transaction leke update kar diya
    
```
